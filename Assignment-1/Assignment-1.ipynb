```python
import numpy as np
import pandas as pd
from numpy.linalg import inv
from sklearn.linear_model import LinearRegression
from sklearn.linear_model import Perceptron
from sklearn.preprocessing import StandardScaler
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
```

```python
# Question-1
```

```python
NUM_ITERATIONS = 5
INIT_WEIGHTS = np.array([0, 6, 6])

samples = np.array([
    [1, 2, 0],  
    [1, 2, 2],
    [1, 0, 2]
])
targets = np.array([1, 1, -1]) 

current_weights = INIT_WEIGHTS.copy()

for iteration in range(NUM_ITERATIONS):
    print(f"\nCycle {iteration + 1}")
    print(f"Current Parameters: {current_weights}")
    
    for idx, observation in enumerate(samples):
        decision_score = np.dot(current_weights, observation)
        predicted_class = 1 if decision_score >= 0 else -1
        
        if predicted_class != targets[idx]:
            current_weights += targets[idx] * observation
            print(f"Sample {idx+1} New weights: {current_weights}")
           
print("\n Final Model Parameters")
print(current_weights)
```

```python
input_features = np.array([
    [1, 3, 2],  
    [1, 2, 2],   
    [1, 1, -1],   
    [1, -1, -3]  
])

target_values = np.array([
    [3],         
    [1],          
    [-1],         
    [-3]          
])

feature_transpose = input_features.T

covariance_matrix = feature_transpose @ input_features
covariance_inverse = np.linalg.inv(covariance_matrix)

projection_components = feature_transpose @ target_values

optimal_weights = covariance_inverse @ projection_components

print("Covariance Matrix (XᵀX):\n", covariance_matrix)
print("\nInverse Covariance Matrix (XᵀX)⁻¹:\n", covariance_inverse)
print("\nProjection Vector (XᵀY):\n", projection_components)
print("\nOptimized Parameters:\n", optimal_weights)
```

```python
class_A_samples = np.array([[4, 3], [2, 2]])     
class_B_samples = np.array([[1, 1], [-2, -1]]) 

centroid_A = np.mean(class_A_samples, axis=0)  
centroid_B = np.mean(class_B_samples, axis=0)  

print("Centroid of Class A:", centroid_A)
print("Centroid of Class B:", centroid_B)


A_deviations = class_A_samples - centroid_A  
B_deviations = class_B_samples - centroid_B  

dispersion_A = np.dot(A_deviations.T, A_deviations)  
dispersion_B = np.dot(B_deviations.T, B_deviations)  

total_dispersion = dispersion_A + dispersion_B

print("\nWithin-Class Scatter Matrix for Class A:\n", dispersion_A)
print("\nWithin-Class Scatter Matrix for Class B:\n", dispersion_B)
print("\nTotal Within-Class Scatter Matrix (Sw):\n", total_dispersion)

class_separation = (centroid_A - centroid_B).reshape(-1, 1) 

projection_params = np.linalg.pinv(total_dispersion) @ class_separation

print("\nOptimal Projection Vector w:\n", projection_params.flatten())
```
